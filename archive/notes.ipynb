```python
from matplotlib import pyplot as plt
import numpy as np
```

```python
size = 150
C = np.zeros((size,size), dtype='int')

C = C+3
C[31:66, 1:-1] = 4
C[32:33, 1:-1] = 5
C[60:62, 1:-1] = 5

C[1:-1, 31:33] = 4
C[1:-1, 63:66] = 4
C[1:-1, 60:62] = 5

C[-3:, 10:] = 4
C[10:, -3:] = 4


```
